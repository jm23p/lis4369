### 1. List all magic commands %lsmagic

```python
%lsmagic
```

```python
# example line magic
%ls
```

### 2. Run file: %run

```python
# example run file
%run hello.py
```

### 3. Load file: %load
**Note:** %run runs file, %load loads file into cell

```python
# %load hello.py
print ("Hello, test!")
# Note: File was loaded into cell, then cell was modified

# create a function
def hello_test():
    print('Hello, test!')

hello_test()
```

### Return execution time: %%time
**Note:** %%time is 'cell magic', and **must be** first line in Jupyter notebook cell

```python
%%time
import random
for i in range(0, 1000000):
    random.random

# display time to generate 1,000,000 random numbers
# Note: Placing these comments above code generates an error
```
